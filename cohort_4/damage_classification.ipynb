```python
engine = create_engine("postgresql+psycopg://datalab:datalab@localhost:5432/nepal")
```

```python
query = """
        SELECT DISTINCT(i.building_id),
                bs.*,
                bd.damage_grade
        FROM id_map i
        JOIN building_structure bs
        ON i.building_id = bs.building_id
        JOIN building_damage bd
        ON bs.building_id = bd.building_id
        WHERE i.district_id = 4
"""
print(query)
```

```python
df = pd.read_sql(query, engine)
df.head()
```

```python
df["damage_grade"] = df.damage_grade.str[-1].astype(int)
```

```python
df["severe_damage"] = (df.damage_grade > 3).astype(int)
```
